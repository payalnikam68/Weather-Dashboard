```python
air_url = "https://api.openweathermap.org/data/2.5/air_pollution"
```

```python
lat = df.loc[0, "Latitude"]
lon = df.loc[0, "Longitude"]

print(lat)
print(lon)
```

```python
### Collect Air Quality Data for All Cities
#Creates an empty list to store air-quality records for all 11 cities.

air_quality_data = []
```

```python
##Fetches air-quality information for each city using its latitude and longitude and stores important pollutants such as AQI, PM2.5, PM10, CO, NO₂, SO₂ and O₃.
for index, row in df.iterrows():

    lat = row["Latitude"]
    lon = row["Longitude"]

    air_params = {
        "lat": lat,
        "lon": lon,
        "appid": API_KEY
    }

    air_response = requests.get(air_url, params=air_params)
    air_data = air_response.json()

    record = {
        "City": row["City"],
        "AQI": air_data["list"][0]["main"]["aqi"],
        "PM2.5": air_data["list"][0]["components"]["pm2_5"],
        "PM10": air_data["list"][0]["components"]["pm10"],
        "CO": air_data["list"][0]["components"]["co"],
        "NO2": air_data["list"][0]["components"]["no2"],
        "SO2": air_data["list"][0]["components"]["so2"],
        "O3": air_data["list"][0]["components"]["o3"]
    }

    air_quality_data.append(record)
```
